```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from scipy.stats import chi2_contingency
```

```python
Click = [512 , 281, 527, 193]
No_click = [24814, 24466, 24349, 25040]

x =pd.DataFrame([Click , No_click], columns=['Version_A', 'Version_B', 'Version_C', 'Version_D'], index =('Click','No_click'))
x
```

```python
alpha = 0.05/6 # using Bonferroni Adjustment
alpha
```

For pairwise comparisons among n groups:

n(n-1)/2

So with 4 groups:

4(4-1)/2 = 6
Bonferroni Adjustment

```python
chisq, pvalue, df, expected = stats.chi2_contingency(x)
```

```python
#alternative
#chisq, pvalue, df, expected = chi2_contingency(x, correction=False)
```

```python
if pvalue < alpha:
  print("We Reject the Null Hypothesis")
else:
  print("We Fail to reject the Null Hypothesis")
```

```python
# calculate the click through rate for each version
ctr_a = x.loc["Click", "Version_A"] / sum(x["Version_A"])
ctr_b = x.loc["Click", "Version_B"] / sum(x["Version_B"])
ctr_c = x.loc["Click", "Version_C"] / sum(x["Version_C"])
ctr_d = x.loc["Click", "Version_D"] / sum(x["Version_D"])

# create a table with all four versions and their click-through rates.
ctrs = pd.DataFrame(
    {"version": ["Version_A", "Version_B", "Version_C", "Version_D"], "CTR": [ctr_a, ctr_b, ctr_c, ctr_d]}
)

ctrs.sort_values("CTR", ascending=False)
```

	•	CTR: C > A > B > D
	•	Drop-off: A > D > C
	•	Homepage-return: D > C > A

The overall chi-square test showed significant differences in CTR across the variants, but post-hoc testing did not identify a single winner based on CTR alone. After considering the additional behavioral metrics, Version A appears to be the best overall option. Although Version C has the highest CTR, it also has the highest drop-off rate, suggesting lower-quality engagement. Version A combines a strong CTR with the lowest drop-off rate, making it the most balanced choice.
